```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
from src.backtest import backtest_model
```

```python
# Load features
df = pd.read_csv("../data/processed_features.csv", index_col=[0,1], parse_dates=[1])

# Load trained XGBoost model
model = joblib.load("../data/best_model_xgb.pkl")
print("✅ Model loaded")
```

```python
portfolio = backtest_model(df, model, top_n=5)
```

```python
plt.figure(figsize=(12,6))
plt.plot(portfolio["Portfolio_Value"], label="Strategy")
plt.plot(portfolio["Benchmark_Value"], label="S&P500")
plt.title("Backtest: Strategy vs Benchmark")
plt.xlabel("Date")
plt.ylabel("Portfolio Value")
plt.legend()
plt.show()

print("Final Portfolio Value:", portfolio["Portfolio_Value"].iloc[-1])
```
